```python
#The numpy.linspace() function is used to generate an array of evenly spaced values between two specified numbers
a = np.linspace(0, 1, num=10)
print(a)
```

```python
#Setting "endpoint=False" excludes the stop value and generates evenly spaced values before it.
b = np.linspace(0, 1, num=10, endpoint=False)
print(b)
```

```python
# Radians
angles = np.array([0, np.pi/2, np.pi])
print(angles)
sin_values = np.sin(angles)

print(sin_values)
```

```python
in_array = np.linspace(-np.pi, np.pi, 10)
out_array = np.sin(in_array)
print(in_array)
print(out_array)
```

```python
plt.plot(in_array, out_array, color = 'red', marker = "o")
plt.title("numpy.sin()")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#np.exp() is a fundamental NumPy function used to calculate the exponential of every element in an input array
in_array = [1, 3, 5]
print ("Input array : ", in_array)

out_array = np.exp(in_array)
print ("Output array : ", out_array)
```

```python
in_array = [1, 1.2, 1.4, 1.6, 1.8, 2]
out_array = np.exp(in_array)

y = [1, 1.2, 1.4, 1.6, 1.8, 2]
plt.plot(in_array, y, color = 'blue', marker = "*")

# red for numpy.exp()
plt.plot(out_array, y, color = 'red', marker = "o")
plt.title("numpy.exp()")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
#numpy.column_stack() function is used to stack 1-D arrays as columns into a 2-D array.It takes a sequence of 1-D arrays and stack them as columns to make a single 2-D array. 
# input array
in_arr1 = np.array(( 1, 2, 3 ))
print ("1st Input array : \n", in_arr1) 

in_arr2 = np.array(( 4, 5, 6 ))
print ("2nd Input array : \n", in_arr2) 

# Stacking the two arrays 
out_arr = np.column_stack((in_arr1, in_arr2))
print ("Output stacked array:\n ", out_arr)
```

```python
arr = [20, 2, 7, 1, 34]
res = np.mean(arr)
print(res)
```

```python
arr = [[14, 17, 12],
       [15,  6, 27],
       [23,  2, 54]]

print(np.mean(arr))           # entire array
print(np.mean(arr, axis=0))   # column-wise mean
print(np.mean(arr, axis=1))   # row-wise mean
```
