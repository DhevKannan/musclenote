```python
import matplotlib.pyplot as plt

# Days of the week
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

# Temperature in Celsius for each day
temperature = [22, 24, 19, 23, 25, 21, 20]

# Create the figure and axes
fig, ax = plt.subplots(figsize=(10, 5))

# Plot the data
ax.plot(days, temperature, marker='o', linestyle='-', color='orange')

# Add title and labels
ax.set_title("Temperature Over a Week")
ax.set_xlabel("Day of the Week")
ax.set_ylabel("Temperature (°C)")

ax.grid(True)

# Show the plot
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# --- Example static data ---
GRAPH_SECONDS = 10
time_values = np.linspace(0, GRAPH_SECONDS, 100)          # 100 time points over 10 seconds
emg_values = np.random.randint(0, 65535, size=100)        # Simulated EMG readings
```

```python
emg_values
```

```python
time_values
```

```python
# Example latest values for note label
raw_value = emg_values[-1]
print(raw_value)
voltage = (raw_value / 65535) * 5
print(voltage)
note = 60  # MIDI note
idx = int(raw_value / 65535 * 15)  # tension index
print(idx)
```

```python
# --- Create figure and axes ---
fig, ax = plt.subplots(figsize=(10, 5))
fig.patch.set_facecolor('#0e1a0e')   # dark green-black background
ax.set_facecolor('#0e1a0e')

# Plot EMG signal
line, = ax.plot(time_values, emg_values, color='#00ff88', linewidth=1.8, label='EMG Signal')

# Labels and title
ax.set_ylim(0, 65535)
ax.set_xlabel('Time (seconds)', color='white', fontsize=12)
ax.set_ylabel('EMG Reading (0 – 65535)', color='white', fontsize=12)
ax.set_title(
    'Static Muscle Signal → C Major Pentatonic Piano\n'
    'Relax = low note (C3) | Squeeze = high note (A5)',
    color='white', fontsize=13, fontweight='bold'
)
ax.tick_params(colors='white')
ax.spines['bottom'].set_color('#336633')
ax.spines['left'].set_color('#336633')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(color='#1a3a1a', linewidth=0.6)
ax.legend(loc='upper right', facecolor='#1a3a1a', labelcolor='white')
```

```python
# Note label
NOTE_NAMES = {48:'C3',50:'D3',52:'E3',55:'G3',57:'A3',
              60:'C4',62:'D4',64:'E4',67:'G4',69:'A4',
              72:'C5',74:'D5',76:'E5',79:'G5',81:'A5'}
note_name = NOTE_NAMES.get(note, str(note))
note_name
```

```python
# --- Create figure and axes ---
fig, ax = plt.subplots(figsize=(10, 5))
fig.patch.set_facecolor('#0e1a0e')   # dark green-black background
ax.set_facecolor('#0e1a0e')

# Plot EMG signal
line, = ax.plot(time_values, emg_values, color='#00ff88', linewidth=1.8, label='EMG Signal')

# Labels and title
ax.set_ylim(0, 65535)
ax.set_xlabel('Time (seconds)', color='white', fontsize=12)
ax.set_ylabel('EMG Reading (0 – 65535)', color='white', fontsize=12)
ax.set_title(
    'Static Muscle Signal → C Major Pentatonic Piano\n'
    'Relax = low note (C3) | Squeeze = high note (A5)',
    color='white', fontsize=13, fontweight='bold'
)
ax.tick_params(colors='white')
ax.spines['bottom'].set_color('#336633')
ax.spines['left'].set_color('#336633')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(color='#1a3a1a', linewidth=0.6) # grid
ax.legend(loc='upper right', facecolor='#1a3a1a', labelcolor='white') #EMG Signal label on upper right

note_label = (
    f"{note_name}  (MIDI {note})  |  {voltage:.2f} V  |  "
    f"{'RELAXED' if idx < 5 else 'MILD' if idx < 10 else 'HIGH TENSION'}"
)
note_text = ax.text(
    0.02, 0.92, note_label,
    transform=ax.transAxes,
    color='#ffcc00', fontsize=14, fontweight='bold',
    bbox=dict(boxstyle='round,pad=0.4', facecolor='#1a3a1a', edgecolor='#336633')
)

plt.show()
```
